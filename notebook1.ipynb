```python
print(df.loc[df['typedouleur'].isin(['A','B'])].head())
print(df.sort_values('age', ascending=False).head())
```

```python
copie=df.copy()
copie.loc[0:3, 'age'] =np.nan
print(copie.isna().sum())
copie['age'] = copie['age'].fillna(copie['age'].median())
```

```python
copie=pd.concat([df,df.iloc[[0]]],ignore_index=True)
print(copie.duplicated().sum())
copie=copie.drop_duplicates()
print(copie.shape)
```

```python
df['age_centre'] = df['age']- df['age'].mean()
df['code_coeur'] = df['coeur'].eq('presence').astype(int)
print(df[['age_centre','code_coeur']].head())
```

```python
print(df.select_dtypes(include='number').apply('mean'))
for col in ['age','tauxmax']:
  print(col, df[col].mean())
```

```python
g = df.groupby('sexe')
print(g[['age','tauxmax']].agg(['mean','std','count']))
```

```python
print(pd.crosstab(df['sexe'], df['coeur']))
print(pd.crosstab(df['sexe'], df['coeur'],normalize='index'))
```
